```python
# A
mstar = u.Msun
v = 30 * u.km / u.s 
Reff = 6 * G * mstar / v**2
print(f"Reff = {Reff.decompose().value:.3e} {Reff.decompose().unit}")

rsolar = 100 * u.AU 

ReffVrsolar = Reff/rsolar 
print(f"ReffVrsolar = {ReffVrsolar.decompose().value} {ReffVrsolar.decompose().unit}")
```

```python
nstar = .1 * (u.pc **(-3))

t = 1/(nstar*(pi *Reff**2) * v)
print(f"t = {t.to(u.yr).value:.3e} {t.to(u.yr).unit}")
```

```python
pcol = (5 *u.Gyr / t).decompose()

print(f"pcol = {pcol.decompose().value:.3e} {pcol.decompose().unit}")
```
