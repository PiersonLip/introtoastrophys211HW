```python
import astropy.units as u
import astropy.constants as const
import numpy as np

h  = const.h        # Planck constant
kB = const.k_B      # Boltzmann constant
c  = const.c        # Speed of light
pi = np.pi
G = const.G
```

# 11.2 

```python
# a

v_esc = 500 * u.km / u.s 
R = 17/2 * u.kpc

M = ((v_esc * np.sqrt(R))/np.sqrt(2 * G))**2

print(f"M = {M.decompose().value:.3e} {M.decompose().unit}")

print(f"M = {M.to(u.Msun).value:.3e} {M.to(u.Msun).unit}")
```

```python
#  b 

m_v = 18.9 
P = 31.47 * u.day

M_v = -2.76 * np.log10(P/(10 * u.day)) - 4.16
print(f"M_v = {M_v.decompose().value:.3e} {M_v.decompose().unit}")

d = 10**((m_v - M_v + 5)/5) * u.pc

print(f"d = {d.decompose().value:.3e} {d.decompose().unit}")
print(f"d = {d.to(u.Mpc).value:.3e} {d.to(u.Mpc).unit}")


# d.decompose()
```

```python
# B ii 

H_0 = 70 * (u.km /u.s)/u.Mpc
rec_v = d *H_0 
rec_v.to(u.km / u.s)
```

```python
# B iii

c_v = -119 * u.km / u.s 
delta_v = abs(rec_v - c_v)
delta_t = abs(14*u.Gyr)

a = delta_v/ delta_t
print(f"a = {a.decompose().value:.3e} {a.decompose().unit}")
```

```python
# C iv 

R = d 

M_MW = (a * R**2/(2*G))

print(f"M_MW = {M_MW.to(u.Msun).value:.3e} {M_MW.to(u.Msun).unit}")
```
