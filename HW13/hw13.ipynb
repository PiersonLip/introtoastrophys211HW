```python
import astropy.units as u
import astropy.constants as const
import numpy as np

h  = const.h        # Planck constant
kB = const.k_B      # Boltzmann constant
c  = const.c        # Speed of light
pi = np.pi
```

# 13.2

```python
H01 = 67 * u.km /u.s / u.Mpc

H02 = 74 * u.km /u.s / u.Mpc
```

```python
# a

d = 112 *u.Mpc 
vH01 = d * H01

vH02 = d * H02


print(f"vH01 = {vH01.decompose().value:.3e} {vH01.decompose().unit}")

print(f"vH02 = {vH02.decompose().value:.3e} {vH02.decompose().unit}")
```

```python
# B

freq = 22.23508 * u.GHz
wavelength = c / freq 

print(f"wavelength = {wavelength.decompose().value:.3e} {wavelength.decompose().unit}")
```

```python
# c 

z1 = vH01/c 

z2 = vH02/c 

print(f"z1 = {z1.decompose().value:.3e} {z1.decompose().unit}")

print(f"z2 = {z2.decompose().value:.3e} {z2.decompose().unit}")

freqobsv1 = freq*(1+z1)
freqobsv2 = freq*(1+z2)

# print(f"freqobsv1 = {freqobsv1.decompose().value:.3e} {freqobsv1.decompose().unit}")
# print(f"freqobsv2 = {freqobsv2.decompose().value:.3e} {freqobsv2.decompose().unit}")

wavelength1 = c/freqobsv1
wavelength2 = c/freqobsv2

print(f"wavelength1 = {wavelength1.decompose().value:.3e} {wavelength1.decompose().unit}")
print(f"wavelength2 = {wavelength2.decompose().value:.3e} {wavelength2.decompose().unit}")

deltawavelength =  wavelength1 - wavelength2

print(f"deltawavelength = {deltawavelength.decompose().value:.3e} {deltawavelength.decompose().unit}")
```

# 13.4

```python
A = np.sqrt(3)/4 *u.km **2
r = 6371 * u.km
thetatot = pi + (A)/(r**2)
print(f"thetatot = {thetatot.decompose().value:.9e} {thetatot.decompose().unit}")

thetaminusp = thetatot - pi 

print(f"thetaminusp = {thetaminusp.decompose().value:.6e} {thetaminusp.decompose().unit}")
```
