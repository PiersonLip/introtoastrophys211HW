```python
import numpy as np
import astropy.units as u
import astropy.constants as const

# given star
T = 43650 * u.K
R = 10.7 * const.R_sun

# threshold (13.6 eV)
E0 = 13.6 * u.eV

# constants
h  = const.h
kB = const.k_B
c  = const.c
pi = np.pi

# dimensionless lower limit
u0 = (E0 / (kB * T)).decompose()           # unitless

# prefactor: 8 π^2 R^2 / c^2 * (kT/h)^3
pref = (8 * pi**2 * R**2 / c**2) * ( (kB*T)/h )**3

# bracket term: e^{-u0} (u0^2 + 2u0 + 2)
bracket = np.exp(-u0.value) * (u0.value**2 + 2*u0.value + 2)

Q_star = (pref * bracket).to(1/u.s)        # photon rate
print(Q_star)
```

```python
n_e = 10**9 * u.m**(-3)
alpha = 2.6 * 10**(-19) * u.m**3 * u.s**(-1)

R_s = (3/(4*pi) * (Q_star)/(n_e**2 * alpha))**(1/3)
R_s.decompose()
print(R_s)
R_s.to(u.pc)
```
