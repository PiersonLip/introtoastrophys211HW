```python
import astropy.units as u
import astropy.constants as const
import numpy as np

h  = const.h        # Planck constant
kB = const.k_B      # Boltzmann constant
c  = const.c        # Speed of light
pi = np.pi
G = const.G
e = np.e
```

# 11.1 

```python
# b 

sigma_0 = 36 * u.Lsun / ( u.pc**2)
h_r = 4 * u.kpc 
R_0 = 8.5 * u.kpc 

MilkwayBrightness = sigma_0 / e**(-R_0/h_r)
MilkwayBrightness
```

```python
# c 
h_r = 4 * u.kpc 

asinf = 2 * pi * MilkwayBrightness *h_r**2

print(f"asinf = {asinf.to(u.Lsun).value:.3e} {asinf.to(u.Lsun).unit}")
```

```python
mwMass = asinf/3 / u.Lsun  * u.Msun

print(f"mwMass = {mwMass.to(u.Msun).value:.3e} {mwMass.to(u.Msun).unit}")
```

```python
# e 

h_r = 4 * u.kpc 
R_0 = 8.5 * u.kpc 

R_0Lum = 2 * pi * MilkwayBrightness * h_r**2 * (1 - (((R_0/(h_r) + 1)))* e **(-R_0/h_r))

print(f"R_0Lum = {R_0Lum.to(u.Lsun).value:.3e} {R_0Lum.to(u.Lsun).unit}")
```
